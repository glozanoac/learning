````python
import io
import zipfile
import requests

RECORD_ID = "24045829"
DEST = "datasets/"   # <- change to your path

# Download and extract
os.makedirs(DEST, exist_ok=True)
url = f"https://zenodo.org/records/{RECORD_ID}/files/CerroWayra_synthetic.zip?download=1"
r = requests.get(url, timeout=300)
r.raise_for_status()
with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    z.extractall(DEST)

root = os.path.join(DEST, "CerroWayra_synthetic")

# List all directories and files recursively, as a tree
for dirpath, dirnames, filenames in os.walk(root):
    dirnames.sort()
    rel = os.path.relpath(dirpath, root)
    level = 0 if rel == "." else rel.count(os.sep) + 1
    indent = "    " * level
    print(f"{indent}{os.path.basename(dirpath)}/")
    for f in sorted(filenames):
        size_kb = os.path.getsize(os.path.join(dirpath, f)) / 1024
        print(f"{indent}    {f}  ({size_kb:,.1f} KB)")
````
